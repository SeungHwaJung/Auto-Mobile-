와인데이터를 사용하여 결정트리로 훈련을
총 6-fold로 교차검증을 사용한 코드를 작성하고,
검증세트에서의 점수를 평균을 내어 저장

```python
#20201259 정승화
import pandas as pd
import numpy as np
from sklearn.datasets import load_wine
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_validate, StratifiedKFold
#와인데이터 준비
wine = pd.read_csv('https://bit.ly/wine_csv_data')
#넘파이배열로 변경
data = wine[['alcohol', 'sugar', 'pH']].to_numpy()
target = wine['class'].to_numpy()
#결정트리 초기화
dt = DecisionTreeClassifier(random_state=42)
#분할기 6-Fold
skfold = StratifiedKFold(n_splits=6, shuffle=True, random_state=42)
#교차검증 validate함수 안에 검증세트 점수 포함
scores = cross_validate(dt, data, target, cv=skfold)
#검증세트 점수 추출
cv_test_scores = scores['test_score']
#검증세트 점수 평균
mean_cv_score = np.mean(cv_test_scores)
```
